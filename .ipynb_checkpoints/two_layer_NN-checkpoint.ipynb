```python
import numpy as np
import pandas as pd
```

```python
handwritten_digits_data = pd.read_csv('train.csv')
```

```python
handwritten_digits_data
# 42000 training examples as rows
# 784 features (pixels for 28 x 28 image) plus one column as a label
```

```python
# Convert pandas dataframe to numpy array 
handwritten_digits_data = np.array(handwritten_digits_data)
```

```python
# Gather dimensions in labels for m rows (examples) and n columns(features + label)
m_examples, n_features_labels = handwritten_digits_data.shape
```

```python
# Shuffle the order of the data 
np.random.shuffle(handwritten_digits_data)
```

```python
# Gather all of the test data
# First 1000 examples after shuffle will be the test set
test_data = handwritten_digits_data[0:1000].T
# All of the correct labels for the test set
y_test = test_data[0]
# All of the features (pixels) for the test set
X_test = test_data[1:n_features_labels] # n = 785
X_test = X_test / 255.0
```

```python
train_data = handwritten_digits_data[1000:m_examples].T
y_train = train_data[0]
X_train = train_data[1:n_features_labels]
X_train = X_train / 255.0
```

```python
# Start by initializing the weights and biases randomly
def init_params():
    # Use He initialization because of first activation layer being ReLU
    W1 = np.random.randn(10, 784) * np.sqrt(2.0 / 784)
    # Initialize biases to zero vector
    b1 = np.zeros((10,1))
    W2 = np.random.randn(10, 10) * np.sqrt(2.0 / 10)
    b2 = np.zeros((10,1))
    return W1, b1, W2, b2

def ReLU(Z):
    return np.maximum(0, Z)

def softmax(Z):
    Z -= np.max(Z, axis = 0)
    A = np.exp(Z) / np.sum(np.exp(Z), axis = 0)
    return A

def ReLU_deriv(Z):
    return (Z>0)

def one_hot(y):
    one_hot_y = np.zeros((y.size, y.max() + 1))
    one_hot_y[np.arange(y.size), y] = 1
    one_hot_y = one_hot_y.T
    return one_hot_y

def forward_prop(W1, b1, W2, b2, X):
    Z1 = W1.dot(X) + b1
    A1 = ReLU(Z1)
    Z2 = W2.dot(A1) + b2
    A2 = softmax(Z2)
    return Z1, A1, Z2, A2

def back_prop(Z1, A1, Z2, A2, W1, W2, X, y):
    one_hot_y = one_hot(y)
    dZ2 = A2 - one_hot_y
    dW2 = (1/m_examples) * dZ2.dot(A1.T)
    db2 = (1/m_examples) * np.sum(dZ2, axis = 1, keepdims=True)
    dZ1 = W2.T.dot(dZ2) * ReLU_deriv(Z1)
    dW1 = (1/m_examples) * dZ1.dot(X.T)
    db1 = (1/m_examples) * np.sum(dZ1, axis = 1, keepdims=True)
    return dW1, db1, dW2, db2

def update_params(W1, b1, W2, b2, dW1, db1, dW2, db2, learning_rate_alpha):
    W1 = W1 - learning_rate_alpha * dW1
    b1 = b1 - learning_rate_alpha * db1
    W2 = W2 - learning_rate_alpha * dW2
    b2 = b2 - learning_rate_alpha * db2
    return W1, b1, W2, b2
```

```python
def get_predictions(A2):
    return np.argmax(A2, 0)

def get_accuracy(predictions, y):
    
    return np.sum(predictions == y) / y.size

def gradient_descent(X, y, alpha, iterations):
    W1, b1, W2, b2 = init_params()
    
    for i in range(iterations + 1):
        Z1, A1, Z2, A2 = forward_prop(W1, b1, W2, b2, X)
        dW1, db1, dW2, db2 = back_prop(Z1, A1, Z2, A2, W1, W2, X, y)
        W1, b1, W2, b2 = update_params(W1, b1, W2, b2, dW1, db1, dW2, db2, alpha)
        # Print out per 20 iterations to keep track of progress
        if i % 20 == 0:
            print("Iteration: ", i)
            predictions = get_predictions(A2)
            print(get_accuracy(predictions, y))

    return W1, b1, W2, b2
    
```

```python
W1, b1, W2, b2 = gradient_descent(X_train, y_train, 0.12, 300)
```

```python
def make_predictions(X, W1, b1, W2, b2):
    _, _, _, A2 = forward_prop(W1, b1, W2, b2, X)
    predictions = get_predictions(A2)
    return predictions
```

```python
test_preds = make_predictions(X_test, W1, b1, W2, b2)
accuracy = get_accuracy(test_preds, y_test)
print("Test accuracy:", accuracy)
```
